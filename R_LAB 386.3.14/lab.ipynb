### What to do when you find missing data?
-   Take note of what columns are missing data
-   Why is it missing? (not always possible, but you should take a look anyway)
-   Decided what to do with missing data:
    -   Drop missing records (`df.dropna()`) - not reccommended you might lose value data
    -   Impute (replace) missing values with a placeholder 'NA', 'Pending'
    -   If ordered data forward or back fill day
    -   Categorical:  Impute with the mode.
    -   Quantitative Data: Impute with mean (normal distribution) or median (skewed/non-normal distribution)

```python
df = df.fillna('None')
```

```python
print('-'*10, 'After Handling Missing Values', '-'*10)
print(df)
```

### Imputing Individual Columns

```python
headers = ['Name', 'Age', 'Weight (kg)', 'Salary']

df = pd.DataFrame(emp_list, columns=headers)
```

```python
print('-'*10, 'Before Handling Missing Values', '-'*10)
print(df)
```

```python
median_weight = df['Weight (kg)'].median()

df['Weight (kg)'] = df['Weight (kg)'].fillna(median_weight)

# Impute salary with an appropriate value
mean_salary = df['Salary'].mean()
df['Salary'] = df['Salary'].fillna(mean_salary)
```

```python
print('-'*10, 'After Handling Missing Values', '-'*10)
print(df)
```

```python
headers = ['Name', 'Age', 'Weight (kg)', 'Salary']

df = pd.DataFrame(emp_list, columns=headers)

print('-'*10, 'Before Handling Missing Values', '-'*10)
print(df)
```

```python
subset = df[['Weight (kg)', 'Salary']].fillna('Pending')
print(subset)
```

```python
mean_salary = df['Salary'].mean()
median_weight = df['Weight (kg)'].median()

df2 = df.fillna(value={'Name': 'Unknown', 'Salary': mean_salary, 'Weight (kg)': median_weight})

print(df2)
```
