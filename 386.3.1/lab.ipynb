```python
import numpy as np
import pandas as pd
```

```python
# Dictionary to make into a data frame
d = {
    'x':[1,2,3],
    'y':[2,4,8],
    'z':100 # scalar values, repeated for all records
}

print(pd.DataFrame(d))
```

```python
# Custom index number using the index argument
# Use columns argument/parameter to select columns or give names if none presents
print(pd.DataFrame(d, index=[100,200,300], columns=['x','y','a']))
```

```python
student_dict = {
    'Name': ['Joe', 'Nat', 'Harry'],
    'Age': [20, 21, 28],
    'Marks': [85.10, 77.80, 91.54]
}

student_df = pd.DataFrame(student_dict)
print(student_df)
```

```python
# .set_index() - allow you to set the index
# set the index as one of the column
student_df = student_df.set_index('Name')
print(student_df)
```

```python
data = {
    'Product': ['Laptop', 'Tablet', 'Phone'],
    'Price': [1200, 300, 800],
    'Quantity': [50,150,100]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Accessing index attribute and using RangeIndex
df.index = pd.RangeIndex(start=1, stop=4, step=1)
print(df)
```

```python
patientID = [101, 102, 103, 104]
name = ['Daniel', 'John', 'Abby', 'Charlie']
dob = ['2023-01-01', '2024-01-02', '3/10/1987 1405', '13th Oct, 2026']

#zip() - function combines lists to make a df, make a tuple of tuples

pt_df = pd.DataFrame(zip(patientID, name, dob), columns=['ID','Name','DOB'])
print(pt_df)
```

```python
stocks = ['IBM', 'APPLE', 'TWITTER', 'GE', 'MSFT']
prices = [115.00, 119.14, 19.77, 25.99, 26]

st_df = pd.DataFrame(zip(stocks, prices), columns=['Stock','Price'])
print(st_df)
```

```python
data = [
    {'x':1, 'y':2, 'z':3},
    {'x':21, 'y':12, 'z':13},
    {'x':13, 'y':25, 'z':37}
]

print(pd.DataFrame(data))
```

```python
data = [
    [1,2,3],
    [2,4,100],
    [3,8,100]
]
print(pd.DataFrame(data, columns=['a','b','c']))
```
