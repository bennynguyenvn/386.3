### Pandas DataTypes
- series(s) - single dimensional array. Single column
    - Homogenous - 1 data type per series
- dataframe(df) - two dimensional array (table)

```python
ser = pd.Series(list('abcdef'))

# give it a name
ser.name = 'alphabet'


# .head(n) - show top 5 records (n = number of records)
print(ser.head(2))

# .tail(n) - show top 5 records (n = number of records)
print(ser.tail(2))
```
