#### R_GLAB 386.3.9 - Adding columns to DataFrames

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Name': ['Jane','Princi','James','Fadi'],
    'Height': [5.1, 6.2, 5.1, 5.2],
    'Qualifications': ['Msc', 'MA', 'Msc', 'Msc'],
    'Score 1': [50, 86, 77, 45],
    'Score 2': [50, 96, 60, 30]
}

df = pd.DataFrame(data)
print(df)
```

```python
address = ['NYC', 'NJ', 'PA', 'CA']

df['Address'] = address
print(df)
```

```python
df['Total Score'] = df['Score 1'] + df['Score 2']
print(df)
```

```python
# Avg score column
df['Avg score'] = (df['Score 1'] + df['Score 2'])/2
print(df)
```

```python
# creating and initializing a list
values = [
    ['Rohan', 455], 
    ['Elvish', 250], 
    ['John', 495],
    ['Sai', 400], 
    ['Eric', 350], 
    ['Adam', 450]
    ]

# Cols: Name, Univ_marks
columns = ['Name', 'Univ_marks']
df = pd.DataFrame(values, columns=columns)
print(df)
```

```python

# Create a column for grade percentage
df['Grade%'] = (df['Univ_marks'] / 500)*100
print(df)
```

```python
age = [20,21,19,25,33,99]
df.insert(1, 'Age', age)
print(df)
```

```python
major = ['Art','Language','Math','Science','Nursing','Business']
df.insert(2, 'Major', major)
print(df)
```

```python
address = ['MI','OH','CA','GA','VA','MD']
df1=df.assign(Address=address)
print(df1)
```
